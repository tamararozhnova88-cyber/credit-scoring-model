# Exploratory Data Analysis — UCI Credit Card Default

Source: supplied UCI dataset. Run from repository root.

```
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
df = pd.read_csv("data/raw/UCI_Credit_Card.csv")
df.shape
```

```
df[["LIMIT_BAL", "AGE", "BILL_AMT1", "PAY_AMT1"]].describe()
```

```
fig, axes = plt.subplots(1, 2, figsize=(12,4))
df["AGE"].hist(bins=30, ax=axes[0]); axes[0].set_title("Age")
df.groupby("PAY_0")["default.payment.next.month"].mean().plot.bar(ax=axes[1]); axes[1].set_title("Default rate by recent payment status")
plt.tight_layout()
```

## Methodology

- Use a stratified 80/20 train/test split with fixed random seed 42.
- Avoid using the test sample for choosing features, algorithms, or hyperparameters.
- Payment-history aggregate features are computed in the sklearn pipeline.
- ROC-AUC selects the model by cross-validation; Precision, Recall and F1 are reported at threshold 0.5.
